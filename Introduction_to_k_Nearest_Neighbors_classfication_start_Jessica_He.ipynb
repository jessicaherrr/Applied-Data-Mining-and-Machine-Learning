# **Introduction to k-Nearest Neighbors (k-NN) Algorithm**

The realm of machine learning and data mining hosts an array of techniques, but among them, the **k-Nearest Neighbors (k-NN) algorithm** stands out due to its sheer simplicity and effectiveness. Often abbreviated as **KNN**, this method has its foundation in the principle of **proximity** and **similarity**.

### **Applications**

The versatility of **KNN** is evident in its wide range of applications, including:

- Determining consumer **credit ratings**
- Evaluating **loan applications**
- Detecting **fraudulent activities**
- Aiding **medical diagnoses**

**KNN** is very popular as a **classifier**, but its application as a **regressor** isn't as prevalent.

## **k-Nearest Neighbors (KNN) Algorithm**

### **Steps:**

1. **Choose the number of neighbors, k:** Decide how many neighbors you want to consider. For instance, if k=3, you'll look at the 3 nearest data points.

2. **Calculate Distances:** For the new data point, calculate the distance (usually Euclidean, but other distances can be used) from this point to all points in the training set.

3. **Sort and Select k-nearest points:** Identify the 'k' training examples that are closest to the point.

4. **Vote (for Classification):**  Assign the class that occurs most frequently among the K neighbors to the new data point.

5. **Average (for Regression):**  Calculate the average of the target values of the K neighbors and assign it as the prediction for the new data point.

### **Advantages:**

- **Simple and intuitive:** No training phase, and easy to explain. KNN doesn't build an explicit model during training. It simply stores the training data in memory, making it readily available for predictions.
- **Versatile:** Can be used for classification, regression, and even recommendation systems.

### **When to Use KNN**

KNN is suitable for various situations:
- When the dataset is not too large, as KNN's performance can degrade with large datasets.
- In cases where there is no prior knowledge about the data distribution.
- For both numeric and categorical data.
- When quick, simple, and interpretable predictions are needed.

### **Disadvantages:**

- **Computationally intensive:** As the dataset grows, calculating distances can be slow.
- **Sensitive to irrelevant features:** Since it relies on distances, irrelevant or redundant features can affect performance.
- **Choice of k:** The value of '**k**' can influence results. A small k can be noisy, while a large k can smooth out decision boundaries too much.

## **Distance Calculation and Metric Selection in k-NN**

The success of the **k-Nearest Neighbors (k-NN)** algorithm hinges on its ability to discern and select the most "similar" instances from a set.

This discernment is achieved through the calculation of distances between data points, which is pivotal in determining the nearest neighbors.



### **Two commonly used distance metrics are:**

#### **1. Euclidean Distance**
- **Euclidean Distance** is a commonly used distance metric that calculates the straight-line distance between two points in a Euclidean space. It is also known as the "**L2 norm**" or "**L2 distance**."
- **Euclidean Distance** is especially useful when dealing with **continuous** and **linear** data spaces.

##### **Equation:**
- The Euclidean Distance between two points $A(x_1, y_1)$ and $B(x_2, y_2)$ in a two-dimensional space can be calculated using the following formula:
  - **Euclidean Distance** $(d) = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$
- Here's a breakdown of the equation:
  - $(x_2 - x_1)$ and $(y_2 - y_1)$ represent the differences in the $x$ and $y$ coordinates between the two points.

#### **2. Manhattan Distance**
- **Manhattan Distance**, also known as the "**L1 norm**" or "**L1 distance**," measures the distance between two points based on the sum of the absolute differences of their coordinates.
- It derives its name from the grid-like layout of streets in cities like Manhattan, where you can only travel along the streets at right angles.

##### **Equation:**
- The Manhattan Distance between two points $A(x_1, y_1)$ and $B(x_2, y_2)$ in a two-dimensional space can be calculated using the following formula:
  - **Manhattan Distance** $(dm) = |x_2 - x_1| + |y_2 - y_1|$

## **K-Nearest Neighbors (KNN) Implementation with Scikit-learn**

Scikit-learn (sklearn) is a widely-used Python library for machine learning. The typical workflow when implementing the K-Nearest Neighbors (KNN) algorithm using scikit-learn includes:

### **1. Data Preprocessing**

- **Feature Matrix & Target Vector**:
  - Extract features (typically denoted as **X**) and target labels (usually denoted as **y**).

### **2. Splitting the Data**

- **Train/Test Split**:
  - Partition data into a training set for building the KNN model and a testing set for evaluation.
  - Use the `train_test_split` function from `model_selection`.

- **Data Scaling**:
  - Essential for KNN, as it's sensitive to feature scales.
  - Fit the scaler on the training data with the `fit()` method.

### **3. Model Creation**

- **Choose Estimator**:
  - Select the KNN estimator (`KNeighborsClassifier` for classification or `KNeighborsRegressor` for regression) based on the task.

- **Hyperparameter Tuning**:
  - Adjust hyperparameters, such as the number of neighbors (**K**), for optimal performance.

### **4. Model Training**

- **Train the Model**:
  - Use the `fit` method with the training data.

### **5. Prediction**

- **Predicting**:
  - Use the `predict` method for classification/regression tasks. - Once trained, use the model to make predictions on the testing data. By comparing these predictions with the actual values, you can assess how well your model is performing.


### **6. Evaluation** : **Calculate Performance Metrics**
   - In the context of **classification** tasks, assess the model's predictions using various metrics. These can include:
     - **Accuracy**: Measures how often the model predicts correctly in classification.
     - **Precision**: Evaluates the proportion of true positive predictions when the model predicts positive.
     - **Recall**: Determines the ability of the model to correctly identify all actual positives.
     - **F1-Score**: Provides a balanced measure of precision and recall, useful for binary classification.

   - In the context of **regression** tasks, assess the model's predictions using various metrics. These can include:
     - **Mean Absolute Error (MAE)**: Measures the average absolute difference between predicted and actual values.
     - **Mean Squared Error (MSE)**: Measures the average squared difference between predicted and actual values.
     - **Root Mean Squared Error (RMSE)**: A square root of MSE, providing an interpretable scale for errors.
     - **R-squared (R2)**: Measures the proportion of the variance in the dependent variable that's predictable from the independent variables.

Before we can train any machine learning model, we need to process the data. This means setting up our feature matrix, often labeled as **X**, and our target vector, labeled as **y**.

**Feature Matrix (**X**):** This matrix contains the data for all the predictor variables. Each column in **X** represents a different feature, and each row represents a different observation.

We need to divide our dataset into two parts: a **training set** and a **testing set**.

- **Testing Set:** Used to evaluate the model's performance on unseen data.

## Hyperparameters and Tuning

A **hyperparameter** is a setting we choose manually *before* training a machine learning model. The model does not learn hyperparameters from the data. Instead, we decide them ahead of time.

For example, in **KNN** we must choose the value of **k** (the number of neighbors). That value is a hyperparameter.

**Hyperparameter tuning** means trying different hyperparameter values to find the combination that gives the best model performance. This is important because the default settings are not always optimal. A small **k** may cause **overfitting**, while a large **k** may cause **underfitting**.

---

## GridSearchCV (Automatic Hyperparameter Tuning)

Instead of manually testing many values, we use **GridSearchCV**, which automatically tests many combinations of hyperparameters and selects the best one. It uses **cross validation (CV)**, which repeatedly splits the training set into smaller parts so the model is tested fairly.

---

## KNN Hyperparameters to Tune

For KNN, the most common hyperparameters to tune are:

- **n_neighbors**: number of neighbors (k)  
- **weights**: uniform voting or distance-based voting  
- **metric**: how distance is calculated (euclidean, manhattan, minkowski)  
- **p**: used only when metric is minkowski (p=1 is manhattan, p=2 is euclidean)


```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'n_neighbors': list(range(2, 21)),
    'metric': ['euclidean', 'manhattan']
}
```

```python
knn = KNeighborsClassifier()
```

```python
grid_search = GridSearchCV(
    estimator=knn,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy'
)
```

```python
grid_search.fit(X_train_scaled, y_train)
```

```python
# Best parameters
grid_search.best_params_
```

```python
# Best CV score
grid_search.best_score_
```

```python
# Extract the best knn
best_knn = grid_search.best_estimator_
```

```python
best_knn
```

```python
y_pred_best = best_knn.predict(X_test_scaled)
```

```python
y_pred_best
```

```python
accuracy_score(y_test, y_pred_best)
```

# **Machine Learning Workflow with Scikit-learn**

## **Model Parameters vs. Hyperparameters**

When working with machine learning algorithms, it's vital to understand the distinction between **model parameters** and **hyperparameters**. These terms often arise in the context of model training and tuning, and understanding the nuances between them is crucial for effective model development.

### **1. Parameters**

### **Definition**:
Parameters are values that the machine learning algorithm determines by itself during the training process. They are internal to the model and are learned from the provided data.

#### **Key Points**:
- **Estimated from Data**: The algorithm itself figures out the optimal values for these parameters.
- **Not Set Manually**: As a user, you don't provide these values; the model does.
- **Example**: In a linear regression model, the weights or coefficients assigned to each feature are considered parameters. The algorithm learns these values during training to best fit the data.

### **2. Hyperparameters**

### **Definition**:
Hyperparameters, on the other hand, are values that are set before the learning process begins. They are external configurations that you, the user, must specify and are not learned from the data. Their settings can influence the model's performance, making them crucial for tuning.

#### **Key Points**:
- **Not Estimated from Data**: These values are set based on prior knowledge, research, or experimentation.
- **Set Manually**: As a user, you decide on the values of these hyperparameters before starting the training.
- **Tuning**: The process of finding the best hyperparameters is known as hyperparameter tuning or optimization.
- **Example**: In the k-nearest neighbors (KNN) algorithm, the number 'k' (indicating how many neighbors to consider) is a hyperparameter. You decide its value, and it's not something the model can learn from the data.

### **2. Handling Categorical Variables with One-Hot Encoding**

Categorical variables like `gender`, and `country_name` usually present non-numeric data. Since most machine learning algorithms prefer numerical input, it's essential to convert these categorical variables into a suitable numeric format.

#### **Why One-Hot Encoding (OHE)?**

OHE is a preferred technique to transmute categorical data into a machine-learning-friendly format. It involves converting each unique category into a new column, where these columns take the value of 1 or 0. Here's the rationale behind its selection:

- **Interpretability**: The outcome of OHE provides binary columns which are straightforward and intuitive.
- **Absence of Order Assumption**: Contrary to techniques like label encoding, OHE doesn't impute an ordinal nature to categories, making it apt for non-ordinal variables.

#### **Implementation using Scikit-Learn**

To achieve this, we'll leverage `OneHotEncoder` from the `sklearn.preprocessing` suite.


```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(drop='first')
```

```python
# Select our categorical columns
cat_columns = df.select_dtypes(include=['object']).drop(columns=['name']).columns
cat_columns
```

```python
# Transform the categorical columns
encoded_features = encoder.fit_transform(df[cat_columns]).toarray()
```

```python
feature_names = encoder.get_feature_names_out(cat_columns)
feature_names
```

```python
encoded_df = pd.DataFrame(encoded_features, columns=feature_names)
encoded_df.head()
```

```python
df.head(2)
```

```python
df = df.drop(columns=cat_columns)
df.head(2)
```

```python
df = pd.concat([df, encoded_df], axis=1)
df.head()
```

**After this process**, the dataset will have **binary columns** for each **categorical value**, with the **first category** in each original column being excluded (due to `drop='first'`). This exclusion helps in avoiding **multicollinearity issues**.

### **Step 4: Splitting the Data into Training and Testing Sets**

When building machine learning models, it's crucial to validate their performance. We can't simply train a model on our entire dataset and hope it'll perform accurately on new, unseen data.

This is where the idea of splitting our dataset into separate training and testing sets comes into play.

- **Training Set**: A subset of the dataset used to train our model. Typically, this comprises a major portion of the dataset, e.g., 70-80%.
  
- **Testing Set**: A subset of the dataset to evaluate our model's performance. This data is unseen during the training phase, providing a way to gauge how our model might perform in real-world, unseen scenarios.

Using the `train_test_split` function from the `sklearn.model_selection` module, we can easily partition our data.

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

By setting the test_size parameter to 0.3, we allocate 30% of the data to the test set and the remaining 70% to the training set. The `random_state` ensures reproducibility in our results.



### **Step 5: Scaling the Features**

#### Conceptual Overview

Feature scaling is a **vital step** in many machine learning algorithms. The primary objective of this process is to standardize the range of independent features in the data. Algorithms that depend on the magnitude of the feature, like **linear regression**, **k-Nearest Neighbors**, and **Support Vector Machines**, can be significantly affected by the scale of the features.

There are multiple methods for feature scaling:

- **StandardScaler**: It standardizes the feature by removing the mean and scaling it to unit variance.
- **MinMaxScaler**: Scales and translates each feature individually such that it's in the given range on the training set, typically between zero and one.
- **RobustScaler**: Uses the median and interquartile range, making it robust to outliers.

For our case, we'll use the `StandardScaler` from `sklearn.preprocessing`.


```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
# Fit and transform on the X_train
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

**Note:** It's important to **fit the scaler only on the training data** and not the test data to avoid **data leakage**. Transforming the test data using the scaler fitted on the training data ensures the model is evaluated on **truly unseen data**.

### **Understanding Evaluation Metrics**

Let's break down each metric and understand what they mean:

#### **Mean Absolute Error (MAE):**

- **MAE** represents the average magnitude of errors between the predicted values and the actual values. It measures how far off, on average, our predictions are from the true values.

- **Results:** A MAE of 4.02 means that, on average, our predictions differ from the actual values by approximately 4.02 units. Smaller MAE values indicate that our model's predictions are closer to the true values, implying better accuracy.

#### **Mean Squared Error (MSE):**

- **MSE:** MSE is similar to MAE but has a slight twist. It calculates the average of the squared differences between the predicted and actual values. Squaring the errors penalizes larger errors more severely than smaller errors.

- **Results:** A MSE of 24.57 means that, on average, the squared errors of our predictions from the actual values amount to 24.57. Like MAE, smaller MSE values indicate that our model's predictions are closer to the true values.

#### **Root Mean Squared Error (RMSE):**

- **RMSE:** RMSE is the square root of MSE. It gives us an error value in the same units as the target variable (price in this case), making it easier to interpret. RMSE is a widely used metric for regression tasks.

- **Results:** An RMSE of 4.96 means that, on average, our predictions have an error of approximately 4.96 units when compared to the true values. Again, lower RMSE values signify better model accuracy.

#### **R-Squared (R2):**

- **R2:** R-squared measures the goodness of fit of our regression model. It provides an indication of how well our model fits the data. R2 ranges from 0 to 1, where 0 indicates that the model explains none of the variance, and 1 indicates that it explains all the variance.

- **Results:** An R-squared value of 0.91 suggests that approximately 91% of the variance in the target variable (price) can be explained by our model. In other words, our model captures 91% of the variability in product prices based on the features we've considered. Higher R2 values indicate that our model provides a better fit to the data.

In summary, these metrics collectively help us evaluate the performance of our regression model. Lower MAE, MSE, and RMSE values indicate better accuracy, while a higher R-squared value signifies a better fit to the data. An R2 of 0.91 is quite good, indicating that our model is explaining a significant portion of the variability in product prices.


## **Understanding Model Coefficients**

Model coefficients provide insight into the relationships between predictors (features) and the target variable. They indicate the change in the target variable for a one-unit change in a predictor, assuming all other predictors are held constant.

For linear regression models, the magnitude and sign of coefficients play a significant role:
- **Magnitude**: The size of the coefficient helps in determining the strength of the relationship.
- **Sign**: A positive sign indicates a positive correlation with the target variable, while a negative sign indicates the opposite.

Let's extract and analyze the coefficients of our model:

1. **Extracting Coefficients**: The `coef_` attribute of the trained model gives the coefficients of the predictors. By extracting them, we can gain a preliminary understanding of the relationships.
   
2. **Mapping Coefficients to Features**: It's essential to associate each coefficient with its corresponding feature for better interpretability. This step provides clarity on which features are positively or negatively influencing our target variable the most.

3. **Sorting by Impact**: We'll sort the features by the magnitude of their coefficients, which will help in understanding the relative importance of each feature in predicting the target variable.


```python
# Extract model coefficients
coefficients = model.coef_
coefficients
```

```python
# List of features names
feature_names = X_train.columns.tolist()
feature_names
```

```python
# Create DataFrame to display coeff with features names
coefficients_df = pd.DataFrame({'Feature': feature_names, 'Coefficient': coefficients})
coefficients_df
```

```python
# Round to 2 decimal places
coefficients_df['Coefficient'] = coefficients_df['Coefficient'].round(2)
coefficients_df
```

```python
# Sort
coefficients_df = coefficients_df.sort_values(by='Coefficient', ascending=False)
coefficients_df
```

# Data Exploration
Data exploration is the crucial initial phase in any data analytics effort, following the definition of the **business goal**. This phase is focused on understanding and preparing the data that will be used in the analysis.

For instance, in the **Cross Industry Standard Process for Data Mining (CRISP-DM)** framework, the process begins with the **Business Understanding** step, which is immediately followed by the **Data Understanding** and **Data Preparation** steps. These steps are essential for setting the stage for effective data mining and machine learning analyses.




image.png

## During the Data Exploration phase, the following activities are typically performed:

- **Loading the Data**: The first step is to load the relevant data into your analysis environment (e.g., a Jupyter notebook) where it can be systematically explored.

- **Exploring the Data**: This involves examining the dataset to gain a sense of its structure and content. Exploration can be done systematically, by reviewing each variable and its values, or cumulatively, by identifying patterns across different variables.

- **Descriptive Statistical Summaries**: Generating statistical summaries, such as means, medians, standard deviations, and distributions, helps in understanding the central tendencies and variability in the data.

- **Visualizations**: Visual tools, such as pair plots, histograms, and box plots, are used to visually inspect relationships between variables, detect anomalies, and identify patterns that may not be obvious from the raw data.

This process of data exploration is vital to fully understanding the typical features of different instances in your dataset, which in turn guides the subsequent steps of data preparation, modeling, and analysis. By thoroughly exploring the data, you can ensure that the models you build are based on a well-understood and appropriately prepared dataset, leading to more accurate and actionable insights.

# Features in the `mpg` Dataset

- **mpg**: Miles per gallon, a measure of fuel efficiency (higher is better).
- **cylinders**: Number of cylinders in the engine (e.g., 4, 6, 8); represents engine design.
- **displacement**: Engine displacement in cubic inches; reflects engine size and capacity.
- **horsepower**: Engine power output in horsepower; higher values indicate more powerful engines.
- **weight**: Vehicle weight in pounds; heavier vehicles typically have lower fuel efficiency.
- **acceleration**: Time (in seconds) to accelerate from 0 to 60 mph; lower values indicate faster acceleration.
- **model_year**: Year of manufacture (e.g., 70 = 1970, 80 = 1980); newer models often have better efficiency.
- **origin**: Region where the car was manufactured (1 = USA, 2 = Europe, 3 = Asia).
- **name**: The name of the car model (e.g., "chevrolet chevelle malibu").

## **3. Visualizing the Data**
- **Visualization** is a powerful tool in data exploration. We'll use **Seaborn** to create various plots that help us visually inspect the relationships between variables and identify patterns.

### **Pair Plot**
- A **pair plot** is useful for visualizing the pairwise relationships between variables in the dataset.


```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a pair plot
sns.pairplot(df)

# Show the plot
plt.show()
```

```python
sns.pairplot(df, corner=True, hue='cylinders', diag_kind='kde')
```

### **Histogram and KDE Plot**
- We can visualize the **distribution** of a single variable, such as `mpg`, using a **histogram** combined with a **KDE plot**.


```python
sns.histplot(df['mpg'], kde=True)

plt.title('Distribution of MPG')
plt.xlabel('MPG')
plt.ylabel('Frequency')

plt.show()
```

- A **histogram** is used to visualize the distribution of the `mpg` variable, showing how frequently different values occur.
- The **KDE plot** overlays a smoothed curve that represents the estimated density of the data, providing a clearer picture of the distribution shape.
- **Adjusting binwidth** helps control the granularity of the histogram bars.


### **Box Plot**
- A **box plot** can be used to identify **outliers** and understand the **spread of values** in the `mpg` variable across different categories, such as the number of cylinders.


The **box plot** also called box-and-whisker plot.

The box plot shows multiple statistical measurements.
- The box has an upper bound of $Q_3$ and a lower bound of $Q_1$, thus allowing us to visualize the interquartile range (IQR).   
- The parallel extending lines from the boxes are called whiskers; they indicate the variability outside the lower and upper quartiles.

- The whiskers extend from the box towards the minimum/maximum.
  - For our purposes, they will extend to $Q_3 + 1.5 \times IQR$ and $Q_1 - 1.5 \times IQR$ and anything beyond will be represented as individual points for outliers

- The horizontal line within the box denotes the median.

- There is also an option to show data outliers, usually as circles or diamonds, past the end of the whiskers.



image.png



```python
sns.boxplot(y=df['mpg'])
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='cylinders', y='mpg', data=df, hue='cylinders');

plt.title('MPG by Number of Cylinders')
plt.xlabel('Cylinders')
plt.ylabel('MPG')

sns.despine()
```

## **Scatter Plot**
Scatter plots are essential tools in data analysis, allowing us to visualize the relationship between two continuous variables. They help in identifying patterns, correlations, and potential outliers in the data.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='horsepower', y='mpg', data=df, hue='cylinders')

plt.title('MPG by Number of Cylinders vs. Horsepower')
plt.xlabel('Horsepower')
plt.ylabel('MPG')

sns.despine()
```

```python
plt.figure(figsize=(10, 6))
sns.regplot(x='horsepower', y='mpg', data=df, ci=None, line_kws={'color':'red', 'linewidth':3})

plt.title('MPG by Number of Cylinders vs. Horsepower')
plt.xlabel('Horsepower')
plt.ylabel('MPG')

sns.despine()
```

## **Bar Plot**
A bar plot is one of the most commonly used visualizations in data analysis. It displays the relationship between a categorical variable and a numeric variable by showing rectangular bars with lengths proportional to the values they represent. Bar plots are particularly useful when you want to compare quantities across different categories.

```python
sns.barplot(x='cylinders', y='mpg', data=df, hue='cylinders');

plt.title('Average MPG by Number of Cylinders')
plt.xlabel('Number of Cylinders')
plt.ylabel('Average MPG')

plt.legend([],[])
sns.despine()
plt.show()
```

```python
# Heat map correlation

corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

### **One-Hot Encoding (OHE)**

One-Hot Encoding converts a **categorical column** into separate numeric columns containing `0` and `1`.

For the `origin` column:

- `usa` → `origin_usa`
- `europe` → `origin_europe`
- `japan` → `origin_japan`

A `1` means the car belongs to that category, and `0` means it does not. The original `origin` column is replaced by these new encoded columns.

```python
df['origin'].unique()
```

```python
df['origin'].map({'usa':0, 'europe':1, 'japan':2})
```

```python
df_encoded = pd.get_dummies(df, columns=['origin'],dtype=int)
df_encoded.head()
```

### **Feature Scaling**

Feature scaling puts numerical features on a similar scale so that variables with large values, such as `weight`, do not dominate variables with smaller values, such as `acceleration`.

A common method is **Standardization**, which transforms each feature so it has approximately:

- Mean = `0`
- Standard deviation = `1`

```python
from sklearn.preprocessing import StandardScaler
```

```python
cols_to_scales = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year']
```

```python
scaler = StandardScaler()
```

```python
df_encoded[cols_to_scales] = scaler.fit_transform(df_encoded[cols_to_scales])
df_encoded.head()
```

```python
df_encoded = df_encoded.drop(['name'], axis=1)
df_encoded.head()
```

## **1. Histogram: Distribution of Horsepower**
### Objective
- Create a histogram to visualize the distribution of the `horsepower` feature.

### Instructions
- Use `sns.histplot()` to create the histogram.
- Customize the number of **bins** to better understand the distribution.
- Add a **KDE (Kernel Density Estimate)** line to smooth the distribution curve.
- Remove the **grid** and unnecessary **spines** for a cleaner look.


```python
sns.histplot(df['horsepower'], bins=20, kde=True)
plt.show()
```

## **2. Scatter Plot: Weight vs. Acceleration**
### Objective
- Create a scatter plot to explore the relationship between `weight` and `acceleration`.

### Instructions
- Use `sns.scatterplot()` to create the scatter plot.
- **Color-code** the points by the `origin` of the car to see if origin affects this relationship.
- Add a **regression line** to visualize the trend.
- Remove the **grid** and unnecessary **spines** for a cleaner look.


```python
sns.scatterplot(x='weight', y='acceleration', data=df, hue='origin')
plt.show()
```

```python
sns.regplot(x='weight', y='acceleration', data=df, ci=None, line_kws={'color':'red', 'linewidth':3})
sns.despine()
plt.show()
```

## **3. Box Plot: Displacement by Origin**
### Objective
- Create a box plot to compare the `displacement` of cars across different `origin` categories.

### Instructions
- Use `sns.boxplot()` to create the box plot.
- Add **notches** to the box plot to indicate the confidence interval around the median.
- Remove the **grid** and unnecessary **spines** for a cleaner look.


```python
sns.boxplot(x='origin', y='displacement', data=df, notch=True)
sns.despine()
plt.show()
```
